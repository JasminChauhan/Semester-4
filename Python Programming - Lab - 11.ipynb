### 06) WAP to generate 100 random lottery tickets and pick two lucky tickets from it and announce them as Winner and Runner up respectively.

```python
import random

def generate_lottery_tickets(count=100):
    """Generate a set of unique lottery tickets."""
    tickets = set()
    while len(tickets) < count:
        ticket = random.randint(100000, 999999) 
        tickets.add(ticket)
    return list(tickets)

def pick_winners(tickets):
    """Pick two lucky tickets: Winner and Runner-up."""
    winner, runner_up = random.sample(tickets, 2)
    return winner, runner_up

tickets = generate_lottery_tickets()

winner, runner_up = pick_winners(tickets)

print("Lottery Results:")
print(f"Winner Ticket: {winner}")
print(f"Runner-up Ticket: {runner_up}")
```

### 11) WAP to demonstrate the use of date time module.

```python
import datetime as dt

# Get the current date and time
current_datetime = dt.datetime.now()
print("Current Date and Time:", current_datetime)

# Get today's date
today = dt.date.today()
print("Today's Date:", today)

# Create a specific date
specific_date = dt.date(2025, 2, 12)
print("Specific Date:", specific_date)

# Convert a string to a date object
date_string = "2025-02-12"
converted_date = dt.datetime.strptime(date_string, "%Y-%m-%d").date()
print("Converted Date from String:", converted_date)

# Calculate the difference between two dates
date_diff = specific_date - today
print("Days until specific date:", date_diff.days)

# Get the current time
current_time = dt.datetime.now().time()
print("Current Time:", current_time)

# Format the date and time
formatted_datetime = current_datetime.strftime("%A, %d %B %Y %I:%M:%S %p")
print("Formatted Date and Time:", formatted_datetime)
```

### 12) WAP to demonstrate the use of the math module.

```python
import math

# 1. Square root of a number
num = 25
print("Square root of", num, "is:", math.sqrt(num))

# 2. Factorial of a number
num2 = 5
print("Factorial of", num2, "is:", math.factorial(num2))

# 3. Power function (x^y)
base = 2
exp = 3
print(f"{base} raised to the power {exp} is:", math.pow(base, exp))

# 4. Value of pi and e
print("Value of Pi:", math.pi)
print("Value of e:", math.e)

# 5. Trigonometric functions
angle = math.radians(30)  # Convert degrees to radians
print("Sine of 30 degrees:", math.sin(angle))
print("Cosine of 30 degrees:", math.cos(angle))
print("Tangent of 30 degrees:", math.tan(angle))

# 6. Logarithm (log base e and log base 10)
num3 = 10
print("Natural Log (ln) of", num3, "is:", math.log(num3))
print("Log base 10 of", num3, "is:", math.log10(num3))

# 7. Ceiling and Floor functions
num4 = 5.7
print("Ceiling of", num4, "is:", math.ceil(num4))
print("Floor of", num4, "is:", math.floor(num4))
```
